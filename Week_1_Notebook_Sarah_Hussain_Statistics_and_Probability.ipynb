```python
#Manual Statistics and Coin flipper
import numpy as np

#Sample data
data = [10,15,20,20,25,30,35]
print("Data:", data)
```

```python
#Mean
def mean_calc(numbers):
  total = 0

  for number in numbers:
    total += number
  mean = total / len(numbers)
  return mean

mean_calc = mean_calc(data)
numpy_mean = np.mean(data)

print("Manual mean:", mean_calc)
print("NumPy mean:", numpy_mean)
```

```python
#Median

def median_calc(numbers):
  sorted_numbers = sorted(numbers)
  values_num = len(sorted_numbers)
  middle = values_num // 2

  if values_num % 2 == 1:
    median = sorted_numbers[middle]
  else:
    first_mid = sorted_numbers[middle - 1]
    second_mid = sorted_numbers[middle]
    median = (first_mid + second_mid) / 2

  return median

my_median = median_calc(data)
numpy_median = np.median(data)

print("Manual median:", my_median)
print("NumPy median:", numpy_median)
```

```python
#Population variance

def variance_calc(numbers):
  mean_calculation = np.mean(numbers)
  squared_diff_tot = 0

  for number in numbers:
    difference = number - mean_calculation
    squared_diff = difference ** 2
    squared_diff_tot += squared_diff

  variance = squared_diff_tot / len(numbers)
  return variance

my_variance = variance_calc(data)
numpy_variance = np.var(data)

print("Manual variance:", my_variance)
print("NumPy variance:", numpy_variance)
```

```python
#Standard deviation

def standard_dev_calc(numbers):
  variance = np.var(numbers)
  standard_dev = variance ** 0.5

  return standard_dev

std_calc = standard_dev_calc(data)
numpy_std = np.std(data)

print("Manual standard deviation:", std_calc)
print("NumPy standard deviation:", numpy_std)
```

```python
#coin flipper
import random

heads = 0
tails = 0

for flip in range(1000):
  result = random.choice(["Heads", "Tails"])

  if result == "Heads":
    heads += 1
  else:
    tails += 1

print("Heads:", heads)
print("Tails:", tails)
```

```python
heads_probability = heads / 1000
tails_probability = tails / 1000

print("Empirical probability of heads:", heads_probability)
print("Empirical probability of tails:", tails_probability)
```

```python
theoretical_probability = 0.5

print("Theoretical probability of heads:", theoretical_probability)
print("Empirical probability of heads:", heads_probability)
```
