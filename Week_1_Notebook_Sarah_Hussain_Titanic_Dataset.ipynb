```python
df = pd.read_csv("titanicdataset.csv")
print("Dataset loaded")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

#1 printing the first 5 rows with df.head()
df.head()
```

```python
#2 selecting specific columns
new_columns = df[["Passengerid","Age", "Fare", "Sex", "2urvived"]]
new_columns.head()
```

```python
#3 only keeping the survivors
survivors = df[df["2urvived"] == 1]
print("Number of survivors:", len(survivors))
survivors.head()
```

```python
#4 filter with 2 conditions
young_high_fare = df[(df["Age"] < 18) & (df["Fare"] > 20)]
young_high_fare[["Passengerid","Age", "Fare", "Sex", "2urvived"]].head()
```

```python
#5 Count missing values with isna() and sum()
missing_values = df.isna().sum()
#missing_values
missing_values[missing_values > 0]
```

```python
#6 Fill missing ages
df_clean = df.copy() #make copy so you don't change original data
median_age = df_clean["Age"].median() #GEt the median age
df_clean["Age"] = df_clean["Age"].fillna(median_age) #fill empty values with median age

print("Median age:", median_age)
print("Missing ages before:", df["Age"].isna().sum())
print("Missing ages after:", df_clean["Age"].isna().sum())
```

```python
#7 Remove duplicate rows
duplicate_count = df_clean.duplicated().sum()
print("Duplicate rows before removal:", duplicate_count)
df_clean = df_clean.drop_duplicates()
print("Rows after removal:", len(df_clean))
```

```python
#8 Survival rate of each sex
survival_by_sex = df_clean.groupby("Sex")["2urvived"].mean() * 100
#Groupby seperates into female and male passenger, "2urvived" selects column
#mean calculates the avg of 0 and 1s --> survival rate
# * 100 to make it percentage
survival_by_sex.round(2)
```

```python
#9 Group by passenger class
class_summary = df_clean.groupby("Pclass").agg(
    passenger_count = ("2urvived", "count"),
    average_age = ("Age", "mean"),
    average_fare = ("Fare", "mean"))
class_summary.round(2)
```

```python
#10 Merge 2 dataframes
class_info = pd.DataFrame({
    "Pclass": [1, 2, 3],
    "class_name": ["First Class", "Second Class", "Third Class"]
})

merged_df = df_clean.merge(class_info, on="Pclass")

merged_df[["Pclass", "class_name", "Sex", "Age", "2urvived"]].head()
#The merge() function combines two DataFrames using their shared pclass column. It adds the class name to each passenger based on their passenger-class number.
```
